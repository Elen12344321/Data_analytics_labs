    - Визначити кількість унікальних працівників у кожному відділі.
    - Представити цю інформацію у впорядкованому вигляді, починаючи з відділу з найбільшою кількістю працівників.

8. **Знайдіть співробітників з декількома посадами**
    
    Деякі співробітники можуть змінювати свої посади протягом роботи в компанії. Визначте всіх співробітників, які мали більше однієї посади за час роботи в компанії. Відобразіть їх `emp_no`, `first_name`, `last_name` та список всіх посад, на яких вони працювали.
    
    **Підказка:** Для цього потрібно зробити join таблиці `titles` до `employees`. Для того аби сконкатенувати кілька назв позицій можна використати агрегаційну функцію GROUP_CONCAT.

```python
cursor = con.cursor()

query = """
SELECT
    e.emp_no,
    e.first_name,
    e.last_name,
    GROUP_CONCAT(DISTINCT t.title ORDER BY t.title SEPARATOR ', ') AS titles
FROM
    employees e
    INNER JOIN titles t ON e.emp_no = t.emp_no
GROUP BY
    e.emp_no,
    e.first_name,
    e.last_name
HAVING
    COUNT(DISTINCT t.title) > 1;
"""

cursor.execute(query)

results = cursor.fetchall()

cursor.close()

df = pd.DataFrame(results, columns=['emp_no', 'first_name', 'last_name', 'titles'])
```

Не забудьте відфільтрувати найновішу дату зарплати для кожного керівника. Тут може бути зручно скористатись конструкцією [WITH](https://dev.mysql.com/doc/refman/8.0/en/with.html) або створити підзапит.

```python
cursor = con.cursor()

query = """
SELECT
    t.dept_name,
    t.emp_no,
    t.first_name,
    t.last_name,
    s.salary
FROM
    (SELECT
        d.dept_name,
        dm.emp_no,
        e.first_name,
        e.last_name,
        MAX(from_date) as from_date,
        MAX(to_date) as to_date
    FROM departments d
    JOIN
        dept_manager dm ON d.dept_no=dm.dept_no
    JOIN
        employees e ON dm.emp_no=e.emp_no
    GROUP BY
        d.dept_name,
        dm.emp_no,
        e.first_name,
        e.last_name
    HAVING
        to_date = '9999-01-01') t
JOIN
    salaries s ON t.emp_no = s.emp_no
                AND s.to_date = '9999-01-01'
GROUP BY
    t.dept_name,
    t.emp_no,
    t.first_name,
    t.last_name,
    s.from_date,
    s.to_date,
    s.salary
HAVING
    s.from_date = MAX(s.from_date)
"""

cursor.execute(query)

results = cursor.fetchall()

cursor.close()

df = pd.DataFrame(results, columns=['dept_name', 'emp_no', 'first_name', 'last_name', 'salary'])
```

3. **Визначення рангу зарплати в межах відділу**

4. **Знаходження зміни зарплати співробітника протягом останніх 2 років**

 Виведіть різницю між поточною зарплатою співробітника з emp_no = 10001 та його зарплатою 2 роки тому. Формат відповіді: номер співробітника, salary_2_years_ago, поточна ЗП, різниця у зарплатах 2 роки тому і зараз.

```python
cursor = con.cursor()

query = """
SELECT s.emp_no,
       LAG(salary, 2) OVER (ORDER BY from_date) as salary_2_years_ago,
       salary,
       salary - LAG(salary, 2) OVER (ORDER BY from_date) as salary_difference
FROM salaries s
WHERE
    s.emp_no = 10001
ORDER BY from_date DESC
LIMIT 1;
"""

cursor.execute(query)

results = cursor.fetchall()

cursor.close()

df = pd.DataFrame(results, columns=['emp_no', 'salary_2_years_ago', 'salary', 'salary_difference'])
```

5. **Визначення найвищого зростання зарплати**

 Знайдіть співробітника, який отримав найбільше збільшення зарплати протягом року, який закінчився 1 рік до найновішої дати в базі даних (максимальне значення from_date в таблиці salaries). Формат відповіді: номер співробітника, ЗП з попереднього року, поточка ЗП, на яку суму змінилась ЗП.

```python
cursor = con.cursor()

query = """
WITH MaxDate AS (
    SELECT MAX(from_date) AS latest_date FROM salaries
),
SalaryDifference AS (
    SELECT s.emp_no,
           LAG(salary) OVER (PARTITION BY
              s.emp_no ORDER BY from_date) AS prev_salary,
           salary,
           salary - LAG(salary) OVER (PARTITION BY
              s.emp_no ORDER BY from_date) AS diff,
           from_date
    FROM salaries s
    JOIN MaxDate
    ON s.from_date BETWEEN DATE_SUB(latest_date, INTERVAL 2 YEAR)
                      AND DATE_SUB(latest_date, INTERVAL 1 YEAR)
)
SELECT emp_no, prev_salary, salary, MAX(diff) AS max_difference
FROM SalaryDifference
GROUP BY
    emp_no,
    prev_salary,
    salary
ORDER BY
    max_difference DESC
LIMIT 1;
"""

cursor.execute(query)

results = cursor.fetchall()

cursor.close()

df = pd.DataFrame(results, columns=['emp_no', 'prev_salary', 'salary', 'max_difference'])
```

6. **Середня зарплата та перший праціник на кожній посаді**

 Для кожної посади виведіть її назву, середню зарплату співробітників на ній, ім'я першого співробітника (або співробітників, якщо в один і той самий день було прийнято на цю посаду кілька людей), який був прийнятий на цю посаду (згідно дати прийому на роботу - hire_date), а також hire_date цього співробітника чи співробітників. У відповіді відсортуйте результати за назвою посади і імʼям співробітника.

```python
cursor = con.cursor()

query = """
WITH AverageSalary AS (
    SELECT
        t.title,
        AVG(s.salary) as avg_salary
    FROM
        titles t
    JOIN
        salaries s ON t.emp_no = s.emp_no
    WHERE
        t.to_date = '9999-01-01' AND s.to_date = '9999-01-01'
    GROUP BY
        t.title
),
EarliestHire AS (
    SELECT
        t.title,
        MIN(e.hire_date) as earliest_hire_date
    FROM
        titles t
    JOIN
        employees e ON t.emp_no = e.emp_no
    WHERE
        t.to_date = '9999-01-01'
    GROUP BY
        t.title
)

SELECT
    a.title,
    a.avg_salary,
    e.first_name,
    eh.earliest_hire_date
FROM
    AverageSalary a
JOIN
    EarliestHire eh ON a.title = eh.title
JOIN
    titles t ON eh.title = t.title
JOIN
    employees e ON t.emp_no = e.emp_no AND eh.earliest_hire_date = e.hire_date
WHERE
    t.to_date = '9999-01-01'
ORDER BY
    title, first_name;

"""

cursor.execute(query)

results = cursor.fetchall()

cursor.close()

df = pd.DataFrame(results, columns=['title', 'prev_salary', 'first_name', 'last_hire_date'])
```

7. **Використання NTILE**
    
    Розподіліть зазначених співробітників на 3 групи за розміром їх поточної зарплати. Виведіть `emp_no`, `first_name`, `last_name`, зарплату та номер групи для кожного з них.
    
    `emp_no` співробітників, зарплати яких будемо досліджувати: `419748, 496734, 264134, 209076, 86631, 456171, 16887, 230255, 246465, 420097, 44362, 280370, 479975, 433320, 473949`

```python
cursor = con.cursor()

query = """
WITH SelectedEmployees AS (
    SELECT
        e.emp_no,
        e.first_name,
        e.last_name,
        s.salary
    FROM
        employees e
    JOIN
        salaries s ON e.emp_no = s.emp_no
    WHERE
        e.emp_no IN (419748, 496734, 264134, 209076, 86631, 456171, 16887, 230255, 246465, 420097, 44362, 280370, 479975, 433320, 473949) AND
        s.to_date = '9999-01-01'
)

SELECT
    emp_no,
    first_name,
    last_name,
    salary,
    NTILE(3) OVER (ORDER BY salary DESC) as salary_group
FROM
    SelectedEmployees;

"""

cursor.execute(query)

results = cursor.fetchall()

cursor.close()

df = pd.DataFrame(results, columns=['emp_no', 'first_name', 'last_name', 'salary', 'salary_group'])
```
