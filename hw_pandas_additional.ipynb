```python
print("Перші та останні 5 записів у колонці 'Timestamp':")

print(data['Timestamp'])
```

```python
import datetime
```

```python
def parse_timestamp(ts):
    try:
        return datetime.datetime.strptime(ts, "%Y-%m-%d %H:%M:%S")
    except ValueError:
        try:
            return datetime.datetime.strptime(ts, "%m/%d/%Y %I:%M %p")
        except ValueError:
            try:
                return datetime.datetime.strptime(ts, "%d.%m.%Y %H:%M")
            except ValueError:
                try:
                    return datetime.datetime.strptime(ts, "%m.%d.%Y %H:%M:%S")
                except ValueError:
                    return pd.NaT

data['Timestamp_converted'] = data['Timestamp'].apply(parse_timestamp)
```

```python
print("\nПерші та останні 5 записів у колонці 'Timestamp_converted' після перетворення:")

print(data['Timestamp_converted'])
```

```python
data['Timestamp'] = data['Timestamp_converted']

print("\nПерші та останні 5 записів у колонці 'Timestamp' після перетворення:")
print(data['Timestamp'])
```

12. Зробіть зводну таблицю, де стовпці - це 'Ваш тайтл', рядки - 'Знання англійської мови', а значення - медіанна ЗП. Які висновки можна зробити?
Тут ще для зручності можна зробити градієнтне забарвлення і відформатувати значення для легшого перегляду.

```python
data['Зарплата у $$$ за місяць, лише ставка після сплати податків'] = data['Зарплата у $$$ за місяць, лише ставка після сплати податків'].astype(float)

data['Знання англійської мови'] = pd.Categorical(data['Знання англійської мови'],
                      categories=['Elementary','Pre-Intermediate','Intermediate','Upper-Intermediate','Advanced'],
                      ordered=True)

data['Ваш тайтл'] = pd.Categorical(data['Ваш тайтл'],
                      categories=['Intern/Trainee', 'Junior', 'Немає тайтлу', 'Middle', 'Senior', 'Tech Lead', 'Team Lead', 'Architect', 'Head', 'Manager'],
                      ordered=True)

pivot_table = data.pivot_table(
    index='Знання англійської мови',
    columns='Ваш тайтл',
    values='Зарплата у $$$ за місяць, лише ставка після сплати податків',
    aggfunc='median'
)

styled_pivot_table = pivot_table.style.background_gradient(cmap='viridis').format("{:.2f}")

styled_pivot_table
```

14. Як зростає зарплата з роками? Для цього нам треба відформатувати колонку "Загальний стаж роботи за спеціальністю" аналогічно до того, як ми це робили для колонки 'Знання англійської мови' і потім вивести медіанні значення для кількості років стажу.

```python
data['Загальний стаж роботи за спеціальністю'] = pd.Categorical(
    data['Загальний стаж роботи за спеціальністю'],
    categories=[
        '6 місяців', '1 рік', '1,5 року', '2 роки', '3 роки', '4 роки',
        '5 років', '6 років', '7 років', '8 років', '9 років', '10 років',
        '11 років', '12 років', '13 років', '14 років', '15 років',
        '20 років', '25 років', '30 років', '35 років', '40 років', '45 років', '50 років'
    ],
    ordered=True
)
```

```python
median_salary_by_experience = data.groupby('Загальний стаж роботи за спеціальністю')['Зарплата у $$$ за місяць, лише ставка після сплати податків'].median()
print(median_salary_by_experience)
```

15. Аналіз ЗП Business Analyst та Data Analyst

```python
ba_salaries = data[data['Ваша посада4'] == 'Business Analyst']['Зарплата у $$$ за місяць, лише ставка після сплати податків']
da_salaries = data[data['Ваша посада4'] == 'Data Analyst']['Зарплата у $$$ за місяць, лише ставка після сплати податків']
```

16. Аналіз часу, коли люди заповнювали форму

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pivot_table = pd.crosstab(data['Основна мова програмування'], data['Ваша спеціалізація'], margins=True)

print(pivot_table)
```

```python
plt.figure(figsize=(14, 8))
sns.heatmap(pivot_table, annot=True, fmt="d", cmap="viridis")
plt.title('Залежність між "Основна мова програмування" та "Ваша спеціалізація"')
plt.xlabel('Ваша спеціалізація')
plt.ylabel('Основна мова програмування')
plt.show()
```

C# / .NET та C++ домінують у різних спеціалізаціях, особливо у .NET та Game Development.
